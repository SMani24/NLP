<div style="direction: ltr; text-align: left;">
<p><b>CBOW (Continuous Bag of Words)</b> is a method/model where the goal is to predict the center word based on the surrounding context words. Basically, we take the sentence, remove the center word, and we want to fill in the blank! To do this, we need to iterate over all the possible windows and create the input (context) / output (target) and add them to our two lists (and convert them to tensors for PyTorch!).</p>

<p><b>Skip-Gram</b> on the other hand is the opposite. Here our goal is to predict the context words given the center word. We basically want to find out which words are likely to appear nearby the center word. Since there are multiple context words, we create pairs for each of them (for example if we have "cat" as the center word and "the" and "sat" as the context, we create two pairs: ("cat", "sat") and ("cat", "the")), add them to our lists and convert them to tensors for PyTorch!</p>
</div>
</div>

Part 1:

```python
import re

def preprocessing_stage_1(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)
    text = text.split()

    return text
```

```python
test_sentence_1 = "Hello World! This is a TEST sentence, with 123 numbers."

print(preprocessing_stage_1(test_sentence_1))
```

Part 2:

```python
def make_singular(word):
    if len(word) <= 3:
        return word
    if word.endswith('ies'):
        return word[:-3] + 'y'
    elif word.endswith('es'):
        return word[:-2]
    elif word.endswith('s') and not word.endswith('ss'):
        return word[:-1]
    else:
        return word

def count_frequency(words, frequency):
    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1
    return frequency

def filter_low_frequency(frequency_dict, min_freq):
    return {word: freq for word, freq in frequency_dict.items() if freq >= min_freq}

def add_special_tokens(special_tokens, frequency_dict):
    for token in special_tokens:
        frequency_dict[token] = float('inf')
    return frequency_dict

def make_mappings(frequency_dict):
    word_to_index = {word: idx for idx, word in enumerate(frequency_dict.keys())}
    index_to_word = {idx: word for word, idx in word_to_index.items()}
    return word_to_index, index_to_word

def preprocessing_stage_2(word_lists, min_freq, special_tokens=['<PAD>', '<UNK>', '<SOS>', '<EOS>']):
    all_words = []
    for sentence in word_lists:
        all_words.extend(sentence)

    for index, word in enumerate(all_words):
        all_words[index] = make_singular(word)

    frequency_dict = dict()
    frequency_dict = add_special_tokens(special_tokens, frequency_dict)
    frequency_dict = count_frequency(all_words, frequency_dict)
    frequency_dict = filter_low_frequency(frequency_dict, min_freq)
    word_to_index, index_to_word = make_mappings(frequency_dict)

    return frequency_dict, word_to_index, index_to_word
```

```python
test_text = [
    "the cat sat on the mat",
    "the dog sat on the log",
    "cats and dogs"
]

preprocessed_corpus = [preprocessing_stage_1(sentence) for sentence in test_text]
final_freq, word_to_idx, idx_to_word = preprocessing_stage_2(preprocessed_corpus, min_freq=2)
print(final_freq)
print(word_to_idx)
```

Part 3:

```python
import torch

def CBOW_collate_fn(batch, context_size):
    contexts = []
    targets = []

    for sentence in batch:
        sentence_length = len(sentence)
        for i in range(sentence_length):
            target = sentence[i]
            context = []
            for j in range(i - context_size, i + context_size + 1):
                if j != i and 0 <= j < sentence_length:
                    context.append(sentence[j])
            if len(context) == 2 * context_size:
                contexts.append(context)
                targets.append(target)

    return torch.tensor(contexts, dtype=torch.long), torch.tensor(targets, dtype=torch.long)

def SkipGram_collate_fn(batch, context_size):
    centers = []
    targets = []

    for sentence in batch:
        sentence_length = len(sentence)
        for i in range(sentence_length):
            center = sentence[i]
            for j in range(i - context_size, i + context_size + 1):
                if j != i and 0 <= j < sentence_length:
                    target = sentence[j]
                    centers.append(center)
                    targets.append(target)

    return torch.tensor(centers, dtype=torch.long), torch.tensor(targets, dtype=torch.long)
```

Part 3:

```python
from torch.utils.data import DataLoader
from functools import partial

def create_dataloader(corpus, batch_size, model_type, context_size):
    if model_type == 'CBOW':
        collate_fn = lambda batch: partial(CBOW_collate_fn, context_size=context_size)(batch)
    elif model_type == 'SkipGram':
        collate_fn = lambda batch: partial(SkipGram_collate_fn, context_size=context_size)(batch)
    else:
        raise ValueError("model_type must be either 'CBOW' or 'SkipGram'")

    dataloader = DataLoader(corpus, batch_size=batch_size, shuffle=True, collate_fn=collate_fn)
    return dataloader

def preprocess(corpus, batch_size, model_type, context_size, min_freq):
    preprocessed_corpus = [preprocessing_stage_1(sentence) for sentence in corpus]
    final_freq, word_to_idx, idx_to_word = preprocessing_stage_2(preprocessed_corpus, min_freq)

    indexed_corpus = []
    for sentence in preprocessed_corpus:
        indexed_sentence = []
        for word in sentence:
            singular_word = make_singular(word)
            if singular_word in word_to_idx:
                indexed_sentence.append(word_to_idx[singular_word])
            else:
                indexed_sentence.append(word_to_idx['<UNK>'])
        indexed_corpus.append(indexed_sentence)

    dataloader = create_dataloader(indexed_corpus, batch_size, model_type, context_size)

    return dataloader, final_freq, word_to_idx, idx_to_word
```

```python
cbow_train_loader, cbow_freq, cbow_word_to_idx, cbow_idx_to_word = preprocess(
    corpus=train['text'],
    batch_size=32,
    model_type='CBOW',
    context_size=3,
    min_freq=5
)
cbow_val_loader, _, _, _ = preprocess(
    corpus=valid['text'],
    batch_size=32,
    model_type='CBOW',
    context_size=3,
    min_freq=5
)
cbow_test_loader, _, _, _ = preprocess(
    corpus=test['text'],
    batch_size=32,
    model_type='CBOW',
    context_size=3,
    min_freq=5
)
```

```python
skipgram_train_loader, skipgram_freq, skipgram_word_to_idx, skipgram_idx_to_word = preprocess(
    corpus=train['text'],
    batch_size=32,
    model_type='SkipGram',
    context_size=3,
    min_freq=5
)
skipgram_val_loader, _, _, _ = preprocess(
    corpus=valid['text'],
    batch_size=32,
    model_type='SkipGram',
    context_size=3,
    min_freq=5
)
skipgram_test_loader, _, _, _ = preprocess(
    corpus=test['text'],
    batch_size=32,
    model_type='SkipGram',
    context_size=3,
    min_freq=5
)
```

## <div style="text-align: center; direction: rtl; font-family: Vazir;">پیاده‌سازی شبکه و آموزش شبکه</div>

First we define the networks. Based on the book, Word2Vec networks are just simple shallow networks, we train them but don't actually care bout their results, instead we take their weight matrices as the word embeddings, effectively turning our one hot representations to dense, information rich vectors.

```python
import torch.nn as nn

class CBOWModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim, context_size):
        super(CBOWModel, self).__init__()
        self.embeddings = nn.Embedding(vocab_size, embedding_dim)
        self.linear1 = nn.Linear(embedding_dim, 128)
        self.activation_function1 = nn.ReLU()
        self.linear2 = nn.Linear(128, vocab_size)

    def forward(self, inputs):
        embeds = self.embeddings(inputs)
        avg_embeds = embeds.mean(dim=1)
        out = self.linear1(avg_embeds)
        out = self.activation_function1(out)
        out = self.linear2(out)
        return out

class SkipGramModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim):
        super(SkipGramModel, self).__init__()
        self.embeddings = nn.Embedding(vocab_size, embedding_dim)
        self.linear1 = nn.Linear(embedding_dim, 128)
        self.activation_function1 = nn.ReLU()
        self.linear2 = nn.Linear(128, vocab_size)

    def forward(self, inputs):
        embeds = self.embeddings(inputs)
        out = self.linear1(embeds)
        out = self.activation_function1(out)
        out = self.linear2(out)
        return out

def train_model(model, train_loader, val_loader, num_epochs, learning_rate, device):
    model.to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)

    train_losses = []
    val_losses = []

    for epoch in range(num_epochs):
        model.train()
        total_loss = 0
        for inputs, targets in train_loader:
            inputs, targets = inputs.to(device), targets.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

        avg_loss = total_loss / len(train_loader)
        train_losses.append(avg_loss)
        print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {avg_loss:.4f}")

        model.eval()
        with torch.no_grad():
            val_loss = 0
            for inputs, targets in val_loader:
                inputs, targets = inputs.to(device), targets.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, targets)
                val_loss += loss.item()
            avg_val_loss = val_loss / len(val_loader)
            val_losses.append(avg_val_loss)
            print(f"Validation Loss: {avg_val_loss:.4f}")

    return model, train_losses, val_losses
```

Now let's train the CBOW model.

```python
cbow_model = CBOWModel(vocab_size=len(cbow_word_to_idx), embedding_dim=100, context_size=3)

trained_cbow_model, cbow_model_train_losses, cbow_model_val_losses = train_model(
    model=cbow_model,
    train_loader=cbow_train_loader,
    val_loader=cbow_val_loader,
    num_epochs=15,
    learning_rate=0.001,
    device=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
)
```

```python
import random
random.seed(24)

def choose_random_words(word_to_idx, idx_to_word, num_words=5):
    vocab_size = len(word_to_idx)
    random_indices = random.sample(range(vocab_size), num_words)
    random_words = [idx_to_word[idx] for idx in random_indices]
    return random_words

random_words = choose_random_words(cbow_word_to_idx, cbow_idx_to_word, num_words=5)
print(random_words)
```

Now we shall define cosine similarity and use it to define a function for finding the most similar words to a given word.

```python
def cosine_similarity(vec1, vec2):
    dot_product = torch.dot(vec1, vec2)
    norm_vec1 = torch.norm(vec1)
    norm_vec2 = torch.norm(vec2)
    return dot_product / (norm_vec1 * norm_vec2)

def find_most_similar_words(model, word, word_to_idx, idx_to_word, top_k=5):
    model.eval()
    if word not in word_to_idx:
        return f"Word '{word}' not in vocabulary."

    word_idx = word_to_idx[word]
    word_vector = model.embeddings.weight[word_idx]

    similarities = {}
    for idx in range(len(word_to_idx)):
        if idx != word_idx:
            other_vector = model.embeddings.weight[idx]
            similarity = cosine_similarity(word_vector, other_vector)
            similarities[idx_to_word[idx]] = similarity.item()

    sorted_similarities = sorted(similarities.items(), key=lambda item: item[1], reverse=True)
    return sorted_similarities[:top_k]
```

Now we will find the 5 most similar words to the chosen words for both models.

```python
cbow_similar_words = {}
skip_gram_similar_words = {}
for word in random_words:
    cbow_similar_words[word] = find_most_similar_words(trained_cbow_model, word, cbow_word_to_idx, cbow_idx_to_word, top_k=5)
    skip_gram_similar_words[word] = find_most_similar_words(trained_skip_gram_model, word, skipgram_word_to_idx, skipgram_idx_to_word, top_k=5)

print("CBOW Similar Words:")
for word, similars in cbow_similar_words.items():
    print(f"{word}: {similars}")

print("\nSkip-Gram Similar Words:")
for word, similars in skip_gram_similar_words.items():
    print(f"{word}: {similars}")
```

And finally plotting them using t-SNE

```python
from sklearn.manifold import TSNE
import numpy as np
import matplotlib.pyplot as plt

def plot_embeddings(model, similar_words_dict, word_to_idx, idx_to_word, model_name):
    model.eval()

    num_plots = len(similar_words_dict)
    num_rows = (num_plots + 2) // 3
    num_cols = 3

    plt.figure(figsize=(num_cols * 6, num_rows * 6))

    for i, (word, similars) in enumerate(similar_words_dict.items()):
        words_to_plot = [word] + [sim[0] for sim in similars]
        embeddings = []

        for w in words_to_plot:
            if w in word_to_idx:
                idx = word_to_idx[w]
                embeddings.append(model.embeddings.weight[idx].detach().cpu().numpy())


        if len(embeddings) < 2:
            print(f"Skipping t-SNE for '{word}' due to insufficient data.")
            continue

        tsne = TSNE(n_components=2, random_state=0, perplexity=min(3, len(embeddings) - 1))
        reduced_embeddings = tsne.fit_transform(np.array(embeddings))

        plt.subplot(num_rows, num_cols, i + 1)
        for j, w in enumerate(words_to_plot):
            plt.scatter(reduced_embeddings[j, 0], reduced_embeddings[j, 1])
            plt.annotate(w, (reduced_embeddings[j, 0], reduced_embeddings[j, 1]))
        plt.title(f"Top similar words for '{word}'")
        plt.grid(True)

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.suptitle(f"{model_name} Word Embeddings Visualization", fontsize=16)
    plt.show()

plot_embeddings(trained_cbow_model, cbow_similar_words, cbow_word_to_idx, cbow_idx_to_word, "CBOW")
plot_embeddings(trained_skip_gram_model, skip_gram_similar_words, skipgram_word_to_idx, skipgram_idx_to_word, "Skip-Gram")
```

<p>Based on some research and reviewing the documentation, I arrived at the following explanation:</p>

<p><b>FastText</b> was developed by Facebook to address some of the shortcomings of Word2Vec. It works by treating words as being composed of smaller parts called <b>character n-grams</b> (as opposed to Word2Vec, which treats whole words as single tokens). FastText assigns vectors (embeddings) to each character n-gram and derives the final word embedding by combining these values. To give a better example:</p>

<p>If we use a size of 3 (trigrams), the word "apple" is represented by:</p>
<ul>
    <li>The whole word: <code>&lt;apple&gt;</code></li>
    <li>The character n-grams: <code>&lt;ap</code>, <code>app</code>, <code>ppl</code>, <code>ple</code>, <code>le&gt;</code></li>
</ul>

<p>This approach has two main advantages that result in us not needing aggressive preprocessing:</p>

<p><b>1. Handling Noise, New Words, and Punctuation:</b><br>
In Word2Vec, if we encountered a word like "apple!", the model would treat it as a totally new word compared to "apple". If "apple!" wasn't in the training data, the model would have no choice but to mark it as Unknown (<code>&lt;UNK&gt;</code>). But in FastText, since we break it down into subwords (<code>app</code>, <code>ple</code>, etc.), the model can handle this new variation and arrive at an embedding very similar to "apple" without needing to have seen "apple!" specifically during training.</p>
<p>This eliminates the need for removing punctuation before training.</p>

<p><b>2. Morphology:</b><br>
Using subwords also helps FastText understand that words like "call", "called", and "calling" are related, and it assigns similar embeddings to them because they share root n-grams. This is a huge help as we no longer need to remove suffixes (stemming) or change the tense during preprocessing!</p>
</div>

```python
random_seed = 24

def sample_data(dataset, balanced=True, total_samples=5000):
    if balanced:
        samples_per_class = total_samples // 4
        sampled_data = []
        for class_label in range(4):
            class_data = dataset.filter(lambda x: x['label'] == class_label)
            sampled_class_data = class_data.shuffle(seed=random_seed).select(range(samples_per_class))
            sampled_data.extend(sampled_class_data)
        return sampled_data
    else:
        return dataset.shuffle(seed=random_seed).select(range(total_samples))

train_data = sample_data(ds['train'], balanced=True, total_samples=5000)
test_data = sample_data(ds['test'], balanced=True, total_samples=2000)

val_data = []
test_data_final = []
ratio = 0.5
for class_label in range(4):
    class_data = [item for item in test_data if item['label'] == class_label]
    split_index = int(len(class_data) * ratio)
    val_data.extend(class_data[:split_index])
    test_data_final.extend(class_data[split_index:])

test_data = test_data_final

print(f"Train samples: {len(train_data)}")
print(f"Validation samples: {len(val_data)}")
print(f"Test samples: {len(test_data)}")
```

We first have to prepare the data for our model using dataloaders

```python
import torch
def create_embedding_dataloader(embeddings, batch_size):
    inputs = [torch.tensor(vec, dtype=torch.float32) for vec, _ in embeddings]
    targets = [label for _, label in embeddings]

    dataset = torch.utils.data.TensorDataset(torch.stack(inputs), torch.tensor(targets, dtype=torch.long))
    dataloader = torch.utils.data.DataLoader(dataset, batch_size=batch_size, shuffle=True)

    return dataloader

train_loader = create_embedding_dataloader(train_embeddings, batch_size=64)
val_loader = create_embedding_dataloader(val_embeddings, batch_size=64)
test_loader = create_embedding_dataloader(test_embeddings, batch_size=64)
```

Before we begin designing and implementing the network, let's first look at the sizes of the input vectors and the output layer (number of classes).

```python
print("Input layer dimension: ", train_embeddings[0][0].shape)

def get_number_of_classes(embeddings):
    labels = set()
    for _, label in embeddings:
        labels.add(label)
    return len(labels)

print("Output layer dimension: ", get_number_of_classes(train_embeddings))
```

Now let's define the model architecture, after extensive tuning (changing the leraning rate, number epocsh, different layers, ...) the following architecture had the best results:

```python
import torch.nn as nn

def MLP(input_dim, output_dim):
    model = nn.Sequential(
        nn.Linear(input_dim, 128),
        nn.ReLU(),
        nn.Linear(128, 256),
        nn.ReLU(),
        nn.Linear(256, 512),
        nn.ReLU(),
        nn.Linear(512, 512),
        nn.ReLU(),
        nn.Linear(512, 256),
        nn.ReLU(),
        nn.Linear(256, 128),
        nn.ReLU(),
        nn.Linear(128, output_dim)
    )
    return model

classifier = MLP(input_dim=300, output_dim=get_number_of_classes(train_embeddings))
print(classifier)
```

And finally train it on the data

```python
def train_model(model, train_loader, val_loader, num_epochs, learning_rate, device):
    model.to(device)
    best_val_loss = float('inf')
    best_model_state = None

    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)

    train_losses = []
    val_losses = []

    train_accuracy = []
    val_accuracy = []

    for epoch in range(num_epochs):
        model.train()
        total_loss = 0
        for inputs, targets in train_loader:
            inputs, targets = inputs.to(device), targets.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

        avg_loss = total_loss / len(train_loader)
        train_losses.append(avg_loss)
        train_accuracy.append((outputs.argmax(dim=1) == targets).float().mean().item())
        print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {avg_loss:.4f}")

        model.eval()
        with torch.no_grad():
            val_loss = 0
            for inputs, targets in val_loader:
                inputs, targets = inputs.to(device), targets.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, targets)
                val_loss += loss.item()
            avg_val_loss = val_loss / len(val_loader)
            val_losses.append(avg_val_loss)
            val_accuracy.append((outputs.argmax(dim=1) == targets).float().mean().item())
            print(f"Validation Loss: {avg_val_loss:.4f}")
            if avg_val_loss < best_val_loss:
                best_val_loss = avg_val_loss
                best_model_state = model.state_dict()

    best_model = MLP(input_dim=300, output_dim=get_number_of_classes(train_embeddings))
    best_model.load_state_dict(best_model_state)
    best_model.to(device)

    return best_model, train_losses, val_losses, train_accuracy, val_accuracy

trained_classifier, classifier_train_losses, classifier_val_losses, classifier_train_accuracy, classifier_val_accuracy = train_model(
    model=classifier,
    train_loader=train_loader,
    val_loader=val_loader,
    num_epochs=100,
    learning_rate=0.0003,
    device=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
)
```

Now we take a look at some metrics and plots on the train and validation dataset to see if the model has been properly trained before doing a final test on the test data.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(classifier_train_losses, label='Train Loss')
plt.plot(classifier_val_losses, label='Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Training and Validation Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(classifier_train_accuracy, label='Train Accuracy')
plt.plot(classifier_val_accuracy, label='Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Training and Validation Accuracy')
plt.legend()
plt.show()
```

It is worth noting that we keep the best model based on the loss on the validation data, so the overfitting we see here won't be effective the model performance

```python
import pandas as pd

from sklearn.metrics import accuracy_score, precision_recall_fscore_support
def evaluate_model(model, data, device, type='weighted'):
    model.to(device)
    model.eval()
    all_preds = []
    all_labels = []

    with torch.no_grad():
        for inputs, label in data:
            inputs = torch.tensor(inputs, dtype=torch.float).to(device)
            label = torch.tensor(label, dtype=torch.long).to(device)

            outputs = model(inputs.unsqueeze(0))
            _, predicted = torch.max(outputs.data, 1)

            all_preds.append(predicted.item())
            all_labels.append(label.item())

    accuracy = accuracy_score(all_labels, all_preds)
    precision, recall, f1, _ = precision_recall_fscore_support(all_labels, all_preds, average=type)

    return accuracy, precision, recall, f1

train_accuracy, train_precision, train_recall, train_f1 = evaluate_model(trained_classifier, train_embeddings, torch.device('cuda' if torch.cuda.is_available() else 'cpu'))
val_accuracy, val_precision, val_recall, val_f1 = evaluate_model(trained_classifier, val_embeddings, torch.device('cuda' if torch.cuda.is_available() else 'cpu'))

metrics_df = pd.DataFrame({
    'Dataset': ['Train', 'Validation'],
    'Accuracy': [train_accuracy, val_accuracy],
    'Precision': [train_precision, val_precision],
    'Recall': [train_recall, val_recall],
    'F1-Score': [train_f1, val_f1]
})
metrics_df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
def plot_confusion_matrix(model, data, device, dataset_name):
    model.to(device)
    model.eval()
    all_preds = []
    all_labels = []

    with torch.no_grad():
        for inputs, label in data:
            inputs = torch.tensor(inputs, dtype=torch.float).to(device)
            label = torch.tensor(label, dtype=torch.long).to(device)

            outputs = model(inputs.unsqueeze(0))
            _, predicted = torch.max(outputs.data, 1)

            all_preds.append(predicted.item())
            all_labels.append(label.item())

    cm = confusion_matrix(all_labels, all_preds)
    print(cm)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title(f'Confusion Matrix - {dataset_name} Set')
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.show()

plot_confusion_matrix(trained_classifier, train_embeddings, torch.device('cuda' if torch.cuda.is_available() else 'cpu'), 'Train')
plot_confusion_matrix(trained_classifier, val_embeddings, torch.device('cuda' if torch.cuda.is_available() else 'cpu'), 'Validation')
```
