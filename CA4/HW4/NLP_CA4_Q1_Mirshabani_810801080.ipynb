✍️ <b>Answer:</b><br>
We have three main columns as stated in the question. Below is their role and analysis:

<b>1. Schema</b><br>
This defines the general structure of our database. It sets the types of each column, assigns their names, and gives the model a general idea about what kind of dataset it is dealing with. It is common practice to write the schema of each dataset for managing it, and it is absolutely necessary here; otherwise, the model really wouldn't know how to interact with it!

<b>2. Question</b><br>
The Question is our goal. It is what the user wants to achieve, written in natural human language. The goal of our model is to understand this intent and generate appropriate queries for it.

<b>3. Query</b><br>
This acts as our "gold" output. It is the correct SQL command that achieves the desired result. Our goal is to fine-tune the model so it can create such results.

<hr style="border-top: 1px dashed #f0ad4e;">

<b>Analysis of Challenges:</b>

* <b>Schema:</b> LLMs are mostly accustomed to human languages, but schemas are far from normal human language! We can see this by taking a look at the example above (<i>CREATE TABLE dishes...</i>). I imagine a major challenge would be to make the model understand this syntax properly, especially given that it is very sensitive; one wrong capitalization can lead to an inappropriate result!
* <b>Question:</b> This part is written in human language, so it should be easier for the model to understand (e.g., <i>"What is the total inventory value..."</i>). Since it states the intent of the user, the most challenging part here would be the correct understanding of the user's intent, though this should be manageable given the extensive pre-training of LLMs.
* <b>Query:</b> This part is again more difficult, both because the model first needs to understand the structure of SQL and then be able to produce error-free code. This is especially challenging because it is very sensitive to capitalization and punctuation (like the semicolon <code>;</code>).

First let's create a funciton that concatinates the question and schema together, seperated by the end of sentence token.

```python
def format_input(row):
    row['input_text'] = row['question'] + " </s> " + row['schema']
    return row

dataset = dataset.map(format_input)
print(dataset['train'][0])
```

Now let's load the model and tokenizer and prepare the dataset for BART. For the model we used facebook/bart-base for a balance between size/speed and capability (suggested by gemeni!)

As it turned out we should be replacing the padding toekns with -100 so the model skips them when evaluating (since we don't want the model to predict padding in the output sql!)

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

checkpoint = "facebook/bart-base"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
model = AutoModelForSeq2SeqLM.from_pretrained(checkpoint)

def preprocess_function(examples):
    inputs = [doc for doc in examples["input_text"]]
    model_inputs = tokenizer(inputs, max_length=512, truncation=True, padding="max_length")

    labels = tokenizer(text_target=examples["query"], max_length=128, truncation=True, padding="max_length")

    labels["input_ids"] = [
        [(l if l != tokenizer.pad_token_id else -100) for l in label] for label in labels["input_ids"]
    ]

    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

tokenized_dataset = dataset.map(preprocess_function, batched=True)
tokenized_dataset.set_format(type="torch", columns=["input_ids", "attention_mask", "labels"])
```

Now let's define the hyperparameters and trainer.

```python
from transformers import Seq2SeqTrainingArguments, Seq2SeqTrainer, DataCollatorForSeq2Seq
import time

data_collator = DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model)

training_args = Seq2SeqTrainingArguments(
    output_dir="./results",
    eval_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    weight_decay=0.01,
    save_total_limit=3,
    num_train_epochs=3,
    predict_with_generate=True,
    fp16=True,
    logging_strategy="epoch",
    report_to="none"
)

trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["dev"],
    tokenizer=tokenizer,
    data_collator=data_collator,
)
```

Now we can finally train the model and report the loss and main hyper parameters

```python
start_time = time.time()
train_result = trainer.train()
end_time = time.time()

execution_time = end_time - start_time

print(f"Execution Time: {execution_time:.2f} seconds")
print(f"Final Loss: {train_result.training_loss:.4f}")

print("Loss per Epoch:")
for log in trainer.state.log_history:
    if "loss" in log and "epoch" in log:
        print(f"Epoch {log['epoch']}: Loss = {log['loss']:.4f}")

print("Hyperparameters:")
print(f"Learning Rate: {training_args.learning_rate}")
print(f"Batch Size: {training_args.per_device_train_batch_size}")
print(f"Weight Decay: {training_args.weight_decay}")
print(f"Epochs: {training_args.num_train_epochs}")
```

As we can see we saw some improvements in the loss which is a good sign!

Now let's define the functions for Raw EM and Normalized EM.

```python
import sqlparse

def compute_raw_em(prediction, reference):
    return prediction == reference

def normalize_sql(query):
    query = query.strip()
    while query.endswith(';'):
        query = query[:-1]

    query = sqlparse.format(query, reindent=True, keyword_case='upper', use_space_around_operators=True)

    query = " ".join(query.split())

    query = query.lower()

    return query

def compute_normalized_em(prediction, reference):
    norm_pred = normalize_sql(prediction)
    norm_ref = normalize_sql(reference)
    return norm_pred == norm_ref
```

✍️ <b>Answer:</b><br>
BART (Bidirectional and Auto-Regressive Transformers) is an encoder-decoder transformer model. We can think of it as the combination of a model like BERT (encoder-only) with GPT (decoder-only). These encoder-decoder models consist of two sections:

<b>1. The Encoder</b><br>
This part is the bidirectional component. It reads the entire input sequence (unlike decoder-only models that read one token at a time) and looks at the words/tokens to both the left and right of each token to attend to them. The goal for this stage is to produce useful encodings (similar to the contextual encodings of BERT) for the decoder section. This section uses self-attention as it is only paying attention to the input data.

<b>2. The Decoder</b><br>
The decoder part is an auto-regressive model, meaning it generates the output tokens (the SQL query in our case) one token at a time, from left to right. For generating the output, transformer blocks with cross-attention are used; the decoder looks at the encoder's output (which captures a full understanding of the input sequence), and based on that and the tokens it has previously generated, it assigns a probability to the next token. We then sample from it to generate the next token. (Basically, an augmented version of decoder-only networks).

<b>Why this structure suits Text-to-SQL:</b><br>
The reason why this structure is good for our task of Text-to-SQL is that, in general, the input and output tokens in encoder-decoder models can be vastly different in terms of size and domain (hence their main usage in translation). This is true for our case, too, where the input is a question written in human language and a schema in a structured format, but the output is the syntax of SQL!

Another advantage is that the <b>encoder</b> portion of the network is much better suited for handling the <b>schema</b>. You often need to look back and forth to properly understand how the database has been formatted (for example, how the parentheses are closed, which columns are in a specific table, etc.), and the bidirectional nature of the encoder allows it to form these complex relations better.

<hr style="border-top: 1px dashed #f0ad4e;">

<b>Raw EM vs. Normalized EM</b><br>

<b>1. Raw EM (Exact Match)</b><br>
This is the simplest way of checking the output of the model. With this method, we compare the generated output of the model with the gold output character by character (A == B). This implies that even a single extra space or mismatch in capitalization would result in a wrong answer. This is, of course, not ideal because there are many variations of SQL that differ only in spacing, punctuation, or capitalization. These variations do not make the command invalid but would be considered incorrect in this approach. Therefore, for SQL, it is best to avoid this and instead use the next method.

<b>2. Normalized EM</b><br>
Here, instead of comparing the results directly, we first "normalize" both the gold and generated output. Normalization refers to performing simple changes to the string that won't affect its functionality but will make the strings more similar. In our specific example, we make everything lowercase, remove extra spaces, and force spaces before and after operators. None of these changes affect the actual syntax or make a command invalid. This provides a fairer assessment of our model because it won't eliminate a result just because it has a simple difference that wouldn't affect functionality.

We don't really care about how the model predicts the words in prefix, so we will make a dataset for GPT-2 (which is a casual LM) that is like (prefix + gold SQL + EOS) and mask the prefix so it wouldn't be validated.

```python
from transformers import AutoTokenizer

tokenizer_gpt = AutoTokenizer.from_pretrained("gpt2")
tokenizer_gpt.pad_token = tokenizer_gpt.eos_token

def preprocess_gpt2_function(examples):
    model_inputs = {"input_ids": [], "attention_mask": [], "labels": []}

    for prefix, query in zip(examples["prefix"], examples["query"]):
        prefix_ids = tokenizer_gpt(prefix, add_special_tokens=False)["input_ids"]
        query_ids = tokenizer_gpt(query + tokenizer_gpt.eos_token, add_special_tokens=False)["input_ids"]

        full_ids = prefix_ids + query_ids
        full_labels = [-100] * len(prefix_ids) + query_ids

        if len(full_ids) > 512:
            full_ids = full_ids[:512]
            full_labels = full_labels[:512]

        model_inputs["input_ids"].append(full_ids)
        model_inputs["attention_mask"].append([1] * len(full_ids))
        model_inputs["labels"].append(full_labels)

    return model_inputs

tokenized_gpt2_dataset = gpt2_dataset.map(
    preprocess_gpt2_function,
    batched=True,
    remove_columns=gpt2_dataset["train"].column_names
)
```

✍️ <b>Answer:</b><br>
GPT-2 is a <b>Decoder-Only</b> transformer model. Unlike Encoder-Only models (like BERT) or Encoder-Decoder models (like BART), GPT-2 is fundamentally <b>autoregressive</b>. This means the network sees the input sequence one token at a time and generates output tokens sequentially from left to right.

<b>Key Architectural Feature: Masked Self-Attention</b><br>
In this structure, the model uses masked self-attention. This constraint ensures that when the model is processing a specific token, it can only "attend" to (look at) the tokens that came <i>before</i> it. It cannot see future tokens.

<b>Applicability to Text-to-SQL:</b><br>
The reason this structure can work for Text-to-SQL—even though it wasn't explicitly built for translation tasks—is that we can frame almost any NLP task as a <b>Sequence Completion</b> (Next-Word Prediction) problem. By formatting the input as a prompt (Context + Question), we can expect the model to "complete" the pattern by generating the appropriate SQL command.

<b>Potential Points of Failure:</b><br>
However, this architecture introduces specific risks for this task:
<ul>
    <li><b>Lack of Bidirectional Context:</b> When reading a database schema, one often needs to look "back and forth" to understand relationships (e.g., how a foreign key in a later table relates to a primary key seen earlier). A unidirectional decoder struggles with this because it cannot attend to "future" tokens to clarify the meaning of current ones.</li>
    <li><b>Single Stream Processing:</b> Unlike Encoder-Decoder models, which explicitly separate the input (human language) and output (SQL) into different processing stages, Decoder-only models treat everything as a single continuous stream. This lack of separation can make them less robust for strict "translation" tasks where the input and output domains are highly dissimilar.</li>
</ul>

    <li><b>For BART:</b> We could explicitly highlight foreign keys or schema relationships in the input string to help the model with schema linking.</li>
    <li><b>For GPT-2:</b> A larger model, more training data, and <b>few-shot prompting</b> (showing examples of valid SQL in the prompt) would likely be necessary to guide the model away from hallucinations.</li>
