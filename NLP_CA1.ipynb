<p dir="rtl" style="line-height: 1.8; text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">

ایمیل از دو بخش تشکیل می‌شود. که با @ از هم جدا می‌شوند. بخش اول (قبل از @) local-part نام دارد و بخش دوم domain.
<br>
منظور از ایمیل معتبر این است که موارد زیر در آن‌ها رعایت شده‌باشند:
<br>
۱. دو بخش ایمیل با یک و تنها یک @ از هم جدا شده‌باشند.
<br>
۲. در local-part هم نام و هم نام خانوادگی شخص وجود داشته باشد.

۳. در local-part تنها حروف انگلیسی، اعداد و کاراکترهای -،_ و . مجاز هستند. همچنین دو نقطه نمی‌توانند پشت هم بیایند.

۴. در بخش domain یک میزبان داریم و یک پسوند. میزبان و پسوند همیشه با یک نقطه از یکدیگر جدا می‌شوند. (میزبان می‌تواند در خود نقطه داشته باشد، اما دو نقطه‌ی متوالی در domain مجاز نیست.)

To validate the email, we first read the file and separate the sections of each line (to extract the name and email).

After that is ready, we use the `is_email_valid` function to check for the validity of the email. This function consists of two main sections:

### 1. Validating the Local Part
Here we use the local `rf` string.
* First, we ensure that we don't have any `..` using the line `rf"(?:(?!{local_allowed}*\.\.{local_allowed}*)"`. This looks for "any number of allowed characters + `..` + any number of allowed characters."
* The important part is we only look for **allowed characters**, which doesn't include `@`. This limits our search to only the local part.
* We then make sure both the `first_name` and `last_name` are present using these two positive lookaheads:
    * `rf"(?={local_allowed}*{first_name})"`
    * `rf"(?={local_allowed}*{last_name})"`
* Finally, we match the actual characters of the local part using `rf"{local_allowed}+)"`.

### 2. Validating the Domain
Here we use a similar logic.
* We check for `..` using `rf"(?:(?!{domain_allowed}*\.\.{domain_allowed}*)"`.
* We check for the existence of a single `.` separating the host and domain, and that the domain has at least two characters, using `rf"{domain_allowed}+\.[A-Za-z]{{2,}})"`.

These two parts are then combined using `pattern = re.compile(rf"^{local}@{domain}$", re.IGNORECASE)` to match the final, complete email string from start (`^`) to end (`$`).

````python
import re
def is_email_valid(email, first_name, last_name):
    first_name = re.escape(first_name)
    last_name = re.escape(last_name)

    local_allowed = r"[A-Za-z0-9._-]"
    domain_allowed = r"[A-Za-z0-9.-]"

    local = (
        rf"(?:(?!{local_allowed}*\.\.{local_allowed}*)"
        rf"(?={local_allowed}*{first_name})"
        rf"(?={local_allowed}*{last_name})"
        rf"{local_allowed}+)"
    )

    domain = (
        rf"(?:(?!{domain_allowed}*\.\.{domain_allowed}*)"
        rf"{domain_allowed}+\.[A-Za-z]{{2,}})"
    )

    pattern = re.compile(rf"^{local}@{domain}$", re.IGNORECASE)
    return bool(pattern.match(email))

valid_emails = []
with open("emails.txt", "r") as file:
    for info in file:
        name, email = info.split(", ")
        first_name, last_name = name.strip().split("=")[-1].strip().split(" ")
        email = email.split("=")[-1].strip()
        if is_email_valid(email, first_name, last_name):
            print(email)
            valid_emails.append(email)
            print("___________________________________")

print(valid_emails)
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">پیاده‌سازی Auto-Correction با Minimum Edit Distance</div>

🎯 <b>خروجی مورد انتظار:</b><br>
- مقدار minimum distance بین جفت کلمه‌های داده‌شده در بخش توضیحات
</p>

In this section we will implement the simple minimum edit distance using dynamic programming (the same way that was thought in the class)

````python
word_pairs = [["Athletic", "Atlantic"], ["London", "Boston"], ["Action", "Compact"], ["", "Sting"]]

def compute_min_edit_distance(source, target):
    m = len(source)
    n = len(target)

    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if source[i - 1] == target[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = min(
                    dp[i - 1][j] + 1,
                    dp[i][j - 1] + 1,
                    dp[i - 1][j - 1] + 1
                )

    return dp[m][n]

for source, target in word_pairs:
    min_edit_distance = compute_min_edit_distance(source, target)
    print(f"Minimum edit distance between '{source}' and '{target}' is: {min_edit_distance}")
````

Here, we first read all the words from the `vocab.txt` file. Then, we split the sentence, and for each word in it, we compare it to all the words in `vocab` and calculate the min-edit distance from them (to find the correct version of the word).

In order to preserve the punctuation, we separate each word from its punctuation using the following regex patterns:

match = re.search(r'(\w*)(\W*)', word)

This method is very inefficient for large vocabulary sizes, but it gets the job done for our case :)

* The commented section is for my debugging to check what each word gets corrected with.

The reason some of the words are corrected incorrectly (like at -> it) is that they are not present in our vocabulary, so they get mismatched.

````python
vocabulary = []
with open("vocab.txt", "r") as file:
    for line in file:
        for correct_word in line.strip().split(", "):
            vocabulary.append(correct_word)

sentence_to_be_corrected = "helo studnts at the universty are wrting ther frst edit distnce algorthm in pythn, and they reely enjy it!"

# for word in sentence_to_be_corrected.split():
#     word_clean = re.sub(r'[^\w]', '', word)
#     min_distance = float('inf')
#     suggested_word = word_clean
#     for correct_word in vocabulary:
#         distance = compute_min_edit_distance(word_clean.lower(), correct_word.lower())
#         if distance < min_distance:
#             min_distance = distance
#             suggested_word = correct_word
#     print(f"Original word: '{word_clean}' - Suggested correction: '{suggested_word}' with edit distance {min_distance}")

# print("-----------------------------------")

def correct_sentence(sentence, vocabulary):
    corrected_sentence = ""
    for word in sentence.split():
        match = re.search(r'(\w*)(\W*)', word)
        word_clean = match.group(1)
        punctuation = match.group(2)
        min_distance = float('inf')
        suggested_word = word_clean
        for correct_word in vocabulary:
            distance = compute_min_edit_distance(word_clean.lower(), correct_word.lower())
            if distance < min_distance:
                min_distance = distance
                suggested_word = correct_word
        corrected_sentence += suggested_word + punctuation + " "
    return corrected_sentence

corrected_sentence = correct_sentence(sentence_to_be_corrected, vocabulary)
print("Corrected Sentence:")
print(corrected_sentence)
````

For this section, I implemented two versions, one where we only use punctuation to separate the tokens (basically a sentence separator, with the pattern = r"\w+|[^\w\s]") and one where tokenize using whitespace and punctuations (to be closer to actual tokenizers, with the pattern).

````python
import re
sentences = [
    "Hello, world! NLP is fun.",
    "That U.S.A. poster-print costs $12.40...",
    "This is a test sentence"
]

def tokenize_sentence(sentence):
    pattern = r"\w+|[^\w\s]"
    tokens = re.findall(pattern, sentence)
    return tokens

def tokenize_sentence_white_space(sentence):
    pattern = r"[\w\s]+|[^\w\s]"
    tokens = re.findall(pattern, sentence)
    return tokens

for sentence in sentences:
    print("Original Sentence:", sentence)
    tokens = tokenize_sentence(sentence)
    print("(Normal) `Found ", len(tokens), "tokens:")
    print(tokens)
    print("--------------------------")
    tokens = tokenize_sentence_white_space(sentence)
    print("(Punctuation only) `Found ", len(tokens), "tokens:")
    print(tokens)
    print("-----------------------------------------------------------")
````

````python
import re
sentences = [
    "Hello, world! NLP is fun.",
    "That U.S.A. poster-print costs $12.40...",
    "This is a test sentence"
]

def tokenize_sentence(sentence):
    abbreviation_pattern = r"(?:[A-Z]\.)+"
    word_pattern = r"\w+"
    punctuation_pattern = r"\?!|\.{3}|!!|--|,,|;;|::|''|\"\"|[^\w\s]"
    number_pattern = r"\d+(?:\.\d+){0,1}"
    price_pattern = rf"\${number_pattern}"
    pattern = rf"{abbreviation_pattern}|{price_pattern}|{number_pattern}|{word_pattern}|{punctuation_pattern}"
    tokens = re.findall(pattern, sentence)
    return tokens

for sentence in sentences:
    print("Original Sentence:", sentence)
    tokens = tokenize_sentence(sentence)
    print("(Normal) `Found ", len(tokens), "tokens:")
    print(tokens)
    print("-----------------------------------------------------------")
````

✍️ <b>پاسخ تشریحی زیربخش اول:</b><br>
These simple rule-based tokenization methods have a few shortcomings, which mostly come from their lack of context. Below are a few of the issues:

1.  **Wrong tokenization of numbers:**
    * **Issue:** Numbers with decimal points or units (such as `$`) often get tokenized wrong. For example, `$12.40` might be separated into `'$'`, `'12'`, `'.'`, and `'40'`, which is clearly wrong and doesn't convey the meaning at all.
    * **Solution:** To solve this, we need to change our tokenization approach to specifically recognize these patterns, which will be discussed later.

2.  **Abbreviations:**
    * **Issue:** Abbreviated words often have `.` in them. Since the tokenizer uses simple regex patterns and doesn't have any previous knowledge about the words, it can't distinguish between the different usages of the `.` character and thus separates abbreviations into multiple tokens.
    * **Solutions:** We could either use abbreviation dictionaries (to detect the abbreviated words) or train a machine learning algorithm to classify the type of `.` in the sentence.
    * **Our Implemented Solution:** Since we wanted a simple solution, I just assumed every abbreviation follows this pattern: `Capital letter + '.'`.

3.  **Hyphenated words:**
    * **Issue:** Another issue is with hyphenated words such as `"poster-print"`. Similar to abbreviations, the simple regex tokenizer doesn't have context and can't distinguish the `'-'` as being part of the word.
    * **Solutions:** This can be solved similarly to abbreviations, but a simpler solution that would solve most cases is to simply include `'-'` in our "accepted characters" list.

4.  **Repeating punctuations:**
    * **Issue:** There are some special punctuations such as `"..."` and `"?!"` that need to be tokenized together (as their combination conveys a different meaning).
    * **Solution:** We can simply add these specific multi-character strings to our list for punctuation.

To solve these issues, we changed the structure of our pattern. By using the `|` (OR) operator, the regex engine checks for the most specific patterns first before falling back to the more general ones. The final pattern structure is:

✍️ <b>پاسخ تشریحی زیربخش سوم:</b><br>
WordPiece is a very important tokenizer and it's most famously used in BERT. Both WordPiece and BPE try to break the text into tokens which are smaller than words (sub-words). The main difference between these two algorithms are how they merge their token.

### 1. BPE
In each step, BPE counts all the adjacent pair of tokens in the corpus and finds the **most frequent pair** and merges them. The algorithm repeats this process until a certain vocabulary size is reached.

### 2. WordPiece
WordPiece takes a smarter approach. Instead of merging the most frequent pairs, it merges the pair that **maximises the likelihood** on the training data. Here's a more detailed breakdown of how WordPiece works:
    1. **Initialization:** It starts with the same vocabulary as BPE (a set of all the characters).
    2. **The `##` prefix:** WordPiece adds a special prefix (`##`) to every token (character at the beginning) that isn't at the beginning of the word.
    3. **Iterations:** This is the step that differs from the BPE. Here instead of just merging the most frequent tokens, it calculates the likelihood of all the pairs in the corpus and merges the two that have the highest value. In a nutshell the likelihood is roughly calculated as `(frequency of AB) / (frequency of A * frequency of B)`. This way if two tokens are just too common and happen to be next to each other lots of times they wouldn't get a high score (as the denominator of the fraction is higher) and only the pairs of tokens that are together in a meaningful and intentional way get merged!

After WordPiece is trained the tokenization of input text is done greedily (like BPE) but in a different way, instead of just repeating the same merges as training phase, WordPiece looks for the **longest possible prefix** of the word that is present in the vocabulary and chooses that.

````python
# White space normalization
cleaned_texts = []
for i in range(len(ds['train'])):
    text = ds['train'][i]['Persian']
    text = text.replace(r'\u200c', ' ')
    normalized_text = ' '.join(text.split())
    cleaned_texts.append(normalized_text)

# Punctuation normalization
for i in range(len(cleaned_texts)):
    text = cleaned_texts[i]
    text = re.sub(r"([?!«»؟.,:])", " \1 ", text)
    cleaned_texts[i] = text

# Character normalization
char_mapping = {
    'ي': 'ی',
    'ك': 'ک',
    'ۀ': 'ه',
    'ؤ': 'و',
    'إ': 'ا',
    'أ': 'ا',
    'آ': 'ا',
    'ۀ': 'ه',
    'ة': 'ه',
}
for i in range(len(cleaned_texts)):
    text = cleaned_texts[i]
    for old_char, new_char in char_mapping.items():
        text = text.replace(old_char, new_char)
    cleaned_texts[i] = text
````

````python
from tokenizers import Tokenizer, models, normalizers, pre_tokenizers, trainers

vocab_size = 4000

pre_tokenizer = pre_tokenizers.Whitespace()
tokenizer = Tokenizer(models.BPE(unk_token="[UNK]"))
tokenizer.pre_tokenizer = pre_tokenizer
trainer = trainers.BpeTrainer(vocab_size=vocab_size, special_tokens=["[UNK]", "[CLS]", "[SEP]", "[PAD]", "[MASK]"])
tokenizer.train_from_iterator(cleaned_texts, trainer=trainer)
sample_sentence = cleaned_texts[24]
output = tokenizer.encode(sample_sentence)
print("Vocabulary Size:", tokenizer.get_vocab_size())
print("Sample Sentence:", sample_sentence)
print("Tokens:", output.tokens)
print("IDs:", output.ids)
````

````python
import random

class N_GramLanguageModel:
    def __init__(self, n, tokenizer):
        self.n = n
        self.ngram_counts = {}
        self.context_counts = {}
        self.tokenizer = tokenizer

    def train(self, corpus):
        for sentence in corpus:
            tokens = self.tokenizer.encode(sentence)
            token_ids = tuple(tokens.ids)
            token_ids = (self.tokenizer.token_to_id("[CLS]"),) * (self.n - 1) + token_ids + (self.tokenizer.token_to_id("[SEP]"),)
            for i in range(len(token_ids) - self.n + 1):
                ngram = token_ids[i:i+self.n]
                context = ngram[:-1]

                self.ngram_counts[ngram] = self.ngram_counts.get(ngram, 0) + 1
                self.context_counts[context] = self.context_counts.get(context, 0) + 1

    def get_ngram_probability(self, ngram):
        ngram = tuple(ngram)
        context = ngram[:-1]
        ngram_count = self.ngram_counts.get(ngram, 0)
        context_count = self.context_counts.get(context, 0)

        if context_count == 0:
            return 0.0
        return ngram_count / context_count
    
    def generate_sentence(self, max_length=100):
        context = [self.tokenizer.token_to_id("[CLS]")] * (self.n - 1)
        sentence = []

        for _ in range(max_length):
            candidates = []
            probabilities = []

            for token_id in range(self.tokenizer.get_vocab_size()):
                ngram = context + [token_id]
                prob = self.get_ngram_probability(ngram)
                if prob > 0:
                    candidates.append(token_id)
                    probabilities.append(prob)

            if not candidates:
                break

            total_prob = sum(probabilities)
            probabilities = [p / total_prob for p in probabilities]

            next_token_id = random.choices(candidates, weights=probabilities)[0]
            if next_token_id == self.tokenizer.token_to_id("[SEP]"):
                break

            sentence.append(next_token_id)
            context = context[1:] + [next_token_id]

        decoded_sentence = self.tokenizer.decode(sentence)
        return decoded_sentence
````

````python
Ns = [2, 4, 8]
for n in Ns:
    print(f"Training {n}-gram model")
    ngram_model = N_GramLanguageModel(n, tokenizer)
    ngram_model.train(cleaned_texts)
    print(f"Generating sentence with {n}-gram model:")
    generated_sentence = ngram_model.generate_sentence(max_length=100)
    print(generated_sentence)
    print("---------------------------------------------------")
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">روش‌های هموارسازی</div>


<p dir="rtl" style="line-height: 1.8; text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
هر کدام از الگوریتم‌های هموارسازی (Smoothing) که در درس خوانده‌اید (Laplace, Interpolation, Backoff) را پیاده‌سازی کنید.
<br>
یک مدل
<span dir="ltr">4-gram</span>
بسازید و با مجموعه داده توکنایزشده فارسی که در بخش اول این سوال آماده کردید، آموزش دهید.
(می‌توانید از مدل
<span dir="ltr">4-gram</span>
آموزش‌یافته قبلی خود استفاده کنید.)
<br>
در الگوریتم Interpolation مقادیر λ را 0.4, 0.3, 0.2, 0.1 در نظر بگیرید. به این صورت:
P​(w∣h3​,h2​,h1​)=0.4P​(w∣h3​,h2​,h1​)+0.3P​(w∣h2​,h1​)+0.2P​(w∣h1​)+0.1P​(w)
<br>
در الگوریتم Backoff مقدار λ را 0.4 در نظر بگیرید.
<br>
سپس با استفاده از هر یک از این روش‌ها، متن‌هایی به طول 100 توکن تولید کنید.
<br>
متن‌های تولیدشده را با یکدیگر مقایسه کنید و تفاوت آن‌ها را از نظر روانی و تنوع کلمات بررسی کنید.
<br>
در نهایت، با استفاده از احتمالات محاسبه‌شده در هر یک از این روش‌ها، معیار Perplexity را روی مجموعه دادگان فارسی validation - که در بخش سوم این سوال آماده کردید - حساب کنید.
