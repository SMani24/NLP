```python

var = [[0.10102744415828, 0.41551729674238, 0.99517573932253]]



for i in range(2):
    x = var[i][0]
    y = var[i][1]
    z = var[i][2]
    new_x = (4 - y - 2 * z) / 7
    new_y = (5 - x - 6 * z) / 2
    new_z = (10 - x - 7 * y) / 2
    
    var.append([new_x, new_y, new_z])

print(var[0])
print(var[1])
print(var[2])
```
