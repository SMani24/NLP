```python
prompt = "<image>\n<|grounding|>Convert the document to markdown."

sampling_param = SamplingParams(
            temperature=0.0,
            max_tokens=8192,
            # ngram logit processor args
            extra_args=dict(
                ngram_size=30,
                window_size=90,
                whitelist_token_ids={128821, 128822},  # whitelist: <td>, </td>
            ),
            skip_special_tokens=False,
        )

extracted_texts = {}
for pdf_file in unselectable_text_pdfs:
    base_name = os.path.splitext(os.path.basename(pdf_file))[0]
    image_files = sorted([f for f in os.listdir('/content/images') if f.startswith(base_name)])
    images = [Image.open(os.path.join('/content/images', img_file)) for img_file in image_files]

    model_input = []

    for image in images:
        model_input.append({
            "prompt": prompt,
            "multi_modal_data": {"image": image}
        })

    model_outputs = llm.generate(model_input, sampling_param)

    print(f'OCR Results for {pdf_file}:')
    for ouput in model_outputs:
        print(ouput.outputs[0].text)
        print('-----------------------------------')

    extracted_texts[pdf_file] = [ouput.outputs[0].text for ouput in model_outputs]
```

### 3.What metadata we should extract?

The choice of metadata depends heavily on how we split the data.

1.  **Whole PDF:** If we feed the whole PDF to the model, the **Title**, **Date**, and perhaps the **Titles of major sections** would be good metadata. However, feeding the whole document is usually not possible due to processing time and the limited context window of our models.
2.  **Notes (تبصره):** Another option is to use *Tabsareh*, but these are often too small. Furthermore, since not all Articles (*ماده*) have them, they are not an ideal standard unit for splitting.

**Conclusion:** We will split the text by **Article (ماده)**.

**Proposed Metadata Schema:**
Based on this split, we should extract and store the following:
* **Source:** PDF Filename & PDF Title.
* **Date:** Date of the PDF (and Article date if distinct).
* **Structure:** Article Number.
* **Context Linking:** Since some articles may still exceed the context window, we will add metadata pointing to the **Previous Chunk** and **Next Chunk**. This ensures that if the model needs to "read around" a specific segment, it can access the full context.

*Note: Extracting this information is feasible using rule-based methods and does not require complex deep learning models.*

### 4.Extracting the metadata

First let's break the text at Article (Made) level.

```python
all_texts = []
for pdf_file, texts in extracted_texts.items():
    all_texts.append((pdf_file, '\n'.join(texts)))
for pdf_file, text in selectable_text_pdfs:
    all_texts.append((pdf_file, text))

print(all_texts[0][1].split('\n')[:5])
```

```python
def split_by_keywords(text, keywords):
    pattern = '|'.join([re.escape(keyword) for keyword in keywords])
    splits = re.split(pattern, text)
    return [split.strip() for split in splits if split.strip()]

def extract_first_date(text):
    pattern = r'\b(\d{4}),(\d{2}),(\d{2})\b'
    match = re.search(pattern, text)
    if match:
        return f"{match.group(1)}-{match.group(2)}-{match.group(3)}"
    else:
        return None

def extract_first_number(text):
    pattern = r'\b\d+\b'
    match = re.search(pattern, text)
    if match:
        return match.group(0)
    else:
        return None

def extract_title(text):
    lines = text.split('\n')
    for line in lines:
        if line.strip():
            return line.strip()
    return "Untitled"

splitted_contents = {}
for pdf_file, text in all_texts:
    splitted_contents[pdf_file] = []
    pdf_file_name = os.path.splitext(os.path.basename(pdf_file))[0]

    file_date = extract_first_date(text)
    title = extract_title(text)

    sections = split_by_keywords(text, ['ماده'])
    for article in sections:
        article_number = extract_first_number(article)
        article_date = extract_first_date(article) or file_date
        splitted_contents[pdf_file].append({
            'file_name': pdf_file_name,
            'title': title,
            'article_number': article_number,
            'article_date': article_date,
            'content': article
        })
```

Now we will split the articles to make sure they are no more than 512 tokens

```python
from transformers import AutoTokenizer, AutoModel

def split_till_token_limit(text, tokenizer, max_tokens=512):
    words = text.split()
    current_chunk = []
    current_length = 0
    chunks = []

    for word in words:
        tokenized_word = tokenizer.encode(word, add_special_tokens=False)
        word_length = len(tokenized_word)

        if current_length + word_length <= max_tokens:
            current_chunk.append(word)
            current_length += word_length
        else:
            chunks.append(' '.join(current_chunk))
            current_chunk = [word]
            current_length = word_length

    if current_chunk:
        chunks.append(' '.join(current_chunk))

    return chunks

tokenizer = AutoTokenizer.from_pretrained('intfloat/multilingual-e5-large')

dataset_contents = {}

for pdf_file, articles in splitted_contents.items():
    for article in articles:
        content = article['content']
        chunks = split_till_token_limit(content, tokenizer, max_tokens=512)
        for i, chunk in enumerate(chunks):
            if pdf_file not in dataset_contents:
                dataset_contents[pdf_file] = []
            dataset_contents[pdf_file].append({
                'file_name': article['file_name'],
                'title': article['title'],
                'article_number': article['article_number'],
                'article_date': article['article_date'],
                'chunk_number': i,
                'content': chunk
            })
```

Now let's use the example code for the multilingual-e5-large model to embed our texts

```python
import torch.nn.functional as F
import torch

from torch import Tensor
from transformers import AutoTokenizer, AutoModel

device = "cuda" if torch.cuda.is_available() else "cpu"

def average_pool(last_hidden_states: Tensor,
                 attention_mask: Tensor) -> Tensor:
    last_hidden = last_hidden_states.masked_fill(~attention_mask[..., None].bool(), 0.0)
    return last_hidden.sum(dim=1) / attention_mask.sum(dim=1)[..., None]

input_texts = []
for pdf_file, chunks in loaded_dataset_contents.items():
    for chunk in chunks:
        input_texts.append(f"passage: {chunk['content']}")

tokenizer = AutoTokenizer.from_pretrained('intfloat/multilingual-e5-large')
model = AutoModel.from_pretrained('intfloat/multilingual-e5-large')
model.to(device)

embeddings_list = []
batch_size = 16
for i in range(0, len(input_texts), batch_size):
    batch_texts = input_texts[i:i+batch_size]
    batch_dict = tokenizer(batch_texts, max_length=512, padding=True, truncation=True, return_tensors='pt')
    batch_dict = {k: v.to(device) for k, v in batch_dict.items()}

    with torch.no_grad():
        outputs = model(**batch_dict)
    embeddings = average_pool(outputs.last_hidden_state, batch_dict['attention_mask'])
    embeddings = F.normalize(embeddings, p=2, dim=1)

    for embedding in embeddings:
        embeddings_list.append(embedding.cpu().numpy())
```

Now let's add the embeddings back to the dataset_contets

```python
i = 0
for pdf_file, chunks in loaded_dataset_contents.items():
    for chunk in chunks:
        chunk['embedding'] = embeddings_list[i]
        i += 1
```

### 1.State

We first need to define a state, this is the information that gets passed around different nodes so it needs to include all the information we need for making decisions.

To do this we need to store original query, rewritten query, intent of the user, metadata filters, retrieved docs, selected docs and the final answer. We also need a variable to show how many times we have retried the search (to not get stuck in an infinite loop). We define these in AgentState class which enharites form the TypedDict class used in LangGraph

```python
from typing import TypedDict, List, Dict, Any

class AgentState(TypedDict):
    original_query: str         
    rewritten_query: str        
    intent: str                 
    metadata_filter: Dict[str, Any]
    retrieved_docs: List[Any]   
    selected_docs: List[Any]    
    final_answer: str           
    retry_count: int            
```

### 2.Rewrite Query Node

For the LLM, we will be using Gemini API as it's free and easy to use. In this stage we need a simple function to take the user query, ask gemini to rewrite it and generate a good response.

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatGoogleGenerativeAI(model="gemini-3-flash-preview")

rewrite_prompt = ChatPromptTemplate.from_template(
    """
    You are an assistant that helps rewrite user queries in Persian for better search results.
    Given the original user query (in Perisan), rewrite it to be more specific and clear, 
    your response should be a concise and improved version of the query considering
    the intent of the user and schema of the database, it is important to note that the
    of the databse are in Persian so your reponse should be in Persian as well.
    
    Database Schema: {schema}

    User Query: {original_query}
    Rewritten Query: 
    """
)

rewrite_chain = rewrite_prompt | llm | StrOutputParser()

def rewrite_query(state):
    schema_str = """
    file_name, title, article_number, article_date, 
    chunk_number, content, embedding
    """
    
    response = rewrite_chain.invoke({
        "original_query": state['original_query'],
        "schema": schema_str
    })
    
    return {"rewritten_query": response}
```

### 3.Classify Intent Node

For the next node we are going to give the query to out LLM and ask it to classify the intent of the question into one of three categories: 1.Small Talk 2.Abuse 3.Law Question.

```python
intent_prompt = ChatPromptTemplate.from_template(
    """
    You are an assistant that classifies user queries.
    Given the user query (in Persian), classify its intent into one of the following categories:
    1. Small Talk
    2. Abuse
    3. Law Question

    If the user is asking general questions, greeting or making casual conversation, classify it as 'Small Talk'.
    If the user is using offensive language, insults or making inappropriate requests, classify it as 'Abuse'.
    If the user is asking about legal matters, regulations, or laws, classify it as 'Law Question'.

    You response should be one of the three categories only ('Small Talk', 'Abuse', 'Law Question').

    User Query: {original_query}
    Classified Intent: 
    """
)

answer_small_talk_prompt = ChatPromptTemplate.from_template(
    """
    You are a friendly assistant for answering law questions but if the user is 
    greeting or initiating small talk responds to it in Persian.
    Given the user query (in Persian), respond appropriately.

    User Query: {original_query}
    Assistant Response: 
    """
)

llm_flash = ChatGoogleGenerativeAI(model="gemini-3-flash-preview")

intent_chain = intent_prompt | llm_flash | StrOutputParser()
answer_small_talk_chain = answer_small_talk_prompt | llm_flash | StrOutputParser()

def classify_intent(state):
    response = ""

    while not response in ['Small Talk', 'Abuse', 'Law Question']:
        response = intent_chain.invoke({
            "original_query": state['original_query']
        })
    
    if response == 'Small Talk':
        answer = answer_small_talk_chain.invoke({
            "original_query": state['original_query']
        })

        return {"intent": response, "final_answer": answer}

    if response == 'Abuse':
        answer = "متاسفانه من نمی‌توانم به درخواست‌های نامناسب پاسخ دهم. لطفاً سوالی مرتبط با قوانین بپرسید."
        return {"intent": response, "final_answer": answer}

    return {"intent": response}
```

### 4.Metadata Extraction Node

Now let's implement the node for extracting the metadata from our query so we can later filter the database based on that.

We can't really filter based on embedding or content or file name. So the only other filterable metadata would be title, article number and date.

```python
record = {
    "file_name": chunk['file_name'],
    "title": chunk['title'],
    "article_number": chunk['article_number'],
    "article_date": chunk['article_date'],
    "chunk_number": chunk['chunk_number'],
    "content": chunk['content'],
    "embedding": chunk['embedding'].tolist()
}
```

```python
from langchain_core.pydantic_v1 import BaseModel, Field
from typing import Optional

def convert_date_to_persian(date_str):
    persian_digits = '۰۱۲۳۴۵۶۷۸۹'
    western_digits = '0123456789'
    translation_table = str.maketrans(western_digits, persian_digits)
    return date_str.translate(translation_table)

class MetadataFilter(BaseModel):
    title: Optional[str] = Field(
        default=None,
        description="The name or title of the law or regualation in Persian, e.g., 'قانون صدور چک'."
    )
    article_number: Optional[str] = Field(
        default=None,
        description="The specific article number within the law or regulation in Persian, e.g., '۱۰', '۲۵'."
    )
    article_date: Optional[str] = Field(
        default=None,
        description="The data associated with the article or law in the format YYYY-MM-DD, e.g., '۱۴۰۲-۰۵-۱۰'. if the day or month is not available, use '۰۰' as placeholder, e.g., '۱۴۰۲-۰۰-۰۰'."
    )

structured_llm_flash = llm_flash.with_structured_output(MetadataFilter)

def extract_metadata(state):
    prompt = f"""
    Extract relevant legal metadata from the following user query in Persian to be used 
    as a database filter. If no relevant metadata is found, return an empty response (None).

    User Query: {state['rewritten_query']}
    """

    result = structured_llm_flash.invoke(prompt)
    filter_dict = {k: v for k, v in result.dict().items() if v is not None}
    if filter_dict.get('article_date'):
        filter_dict['article_date'] = convert_date_to_persian(filter_dict['article_date'])

    return {"metadata_filter": filter_dict}
```

<h4 dir="rtl" style="font-family: Vazir; width: 85%;">صداقت علمی:</h4> <ul dir="rtl" style="font-family: Vazir; width: 85%; font-size: 16px;"> <li>ما نوت‌بوک‌های تعداد مشخصی از دانشجویان که به صورت تصادفی انتخاب می‌شوند، بررسی خواهیم کرد. این بررسی‌ها اطمینان حاصل می‌کنند که کدی که نوشتید واقعاً پاسخ‌های موجود در نوت‌بوک شما را تولید می‌کند. اگر پاسخ‌های صحیح را در نوت‌بوک خود بدون کدی که واقعاً آن پاسخ‌ها را تولید کند تحویل دهید، این یک مورد جدی از عدم صداقت علمی محسوب می‌شود.</li> <li>ما همچنین بررسی‌های خودکاری را برای تشخیص سرقت علمی در نوت‌بوک‌های کولب انجام خواهیم داد. کپی کردن کد از دیگران نیز یک مورد جدی از عدم صداقت علمی محسوب می‌شود.</li> </ul>
<h4 dir="rtl" style="font-family: Vazir; width: 85%;">توضیحات تکمیلی:</h4> <ul dir="rtl" style="font-family: Vazir; width: 85%; font-size: 16px;">
<li>
خوانایی و دقت بررسی‌ها در گزارش نهایی از اهمیت ویژه‌ای برخوردار است. به تمرین‌هایی که به صورت کاغذی تحویل داده شوند یا به صورت عکس در سایت بارگذاری شوند، ترتیب اثری داده نخواهد شد.</li>
<li>
 همه‌ی کدهای پیوست گزارش بایستی قابلیت اجرای مجدد داشته باشند. در صورتی که برای اجرا مجدد آن‌ها نیاز به تنظیمات خاصی می‌باشد، بایستی تنظیمات مورد نیاز را نیز در گزارش خود ذکر کنید.  دقت کنید که  تمامی کدها باید توسط شما اجرا شده باشند و نتایج اجرا در فایل کدهای ارسالی مشخص باشد. به کدهایی که نتایج اجرای آن‌ها در فایل ارسالی مشخص نباشد نمره‌ای تعلق نمی‌گیرد.
