- <b>LanceDB</b>: برای ذخیره‌سازی و بازیابی هوشمند اطلاعات با RAG

First let's load the dataset.

````python
import kagglehub

path = kagglehub.dataset_download("zinovadr/iata-airport-code")

print("Path to dataset files:", path)

df = pandas.read_csv(os.path.join(path, "airport-codes_csv.csv"))
````

Now let's take a look at the dataset

````python
df.head()
````

Based on this we can see that the column named 'muicipality' contains the city names and the column named 'iata_code' has the codes of the airport code so we will be working with these two columns for creating the mapping.

````python
df.info()
````

````python
df = df[df["iata_code"].notna()]
print("Number of non-null airport codes:", df["iata_code"].notna().sum())
````

However as we can see there are quite a few missing codes, in fact there are only '9,199' non-null airport codes. So we will only be using these for making the mapping.

````python
city_airport_mapping = {}
for index, row in df.iterrows():
    city = row["municipality"]
    airport_code = row["iata_code"]
    if city not in city_airport_mapping:
        city_airport_mapping[city] = []
    city_airport_mapping[city].append(airport_code)
````
